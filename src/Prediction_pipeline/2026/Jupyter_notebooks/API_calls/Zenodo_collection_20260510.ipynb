```python
import json
import os
import time
import requests

BASE_URL = "https://zenodo.org/api/records"
QUERY = "arabidopsis"
SIZE = 25
OUT_DIR = "zenodo_arabidopsis_metadata"

def fetch_page(page: int) -> dict:
    r = requests.get(
        BASE_URL,
        params={"q": QUERY, "size": SIZE, "page": page},
        headers={"Accept": "application/json"},
        timeout=60,
    )
    if not r.ok:
        raise RuntimeError(f"HTTP {r.status_code} for {r.url}\n{r.text}")
    return r.json()

def main():
    os.makedirs(OUT_DIR, exist_ok=True)

    page = 1
    saved = 0

    while True:
        data = fetch_page(page)
        hits = data.get("hits", {}).get("hits", [])
        if not hits:
            break

        for rec in hits:
            rec_id = rec.get("id")
            out_path = os.path.join(OUT_DIR, f"record_{rec_id}.json")

            with open(out_path, "w", encoding="utf-8") as f:
                json.dump(rec, f, ensure_ascii=False, indent=2)

            saved += 1

        print(f"Page {page}: saved {len(hits)} files (total {saved})")
        page += 1
        time.sleep(0.2)  # be polite

    print(f"Done. Saved {saved} JSON files in '{OUT_DIR}/'")

if __name__ == "__main__":
    main()
```
